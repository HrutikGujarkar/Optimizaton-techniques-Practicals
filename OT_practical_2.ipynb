```python
import sympy as sp
```

```python
x, y = sp.symbols('x y')
```

```python
f = x**2 + y**2 - 4*x - 6*y + 13
```

```python
print("Function")
print(f)
```

```python
fx = sp.diff(f, x)
fy = sp.diff(f, y)
```

```python
print("\nPartial Derivative wrt x")
print(fx)
```

```python
print("\nPartial Derivative wrt y")
print(fy)
```

```python
critical_point = sp.solve((fx, fy), (x, y))
```

```python
print("\nStationary Point:")
print(critical_point)
```

```python
fxx = sp.diff(fx, x)
fyy = sp.diff(fy, y)
fxy = sp.diff(fx, y)
```

```python
D = fxx*fyy - fxy**2
```

```python
print("\n Second Partial Derevative")
print("fxx:",fxx)
print("fyy:",fyy)
print("fxy:",fxy)
```

```python
print("Hessian Determinant:",D)
```

```python
if D > 0 and fxx >0:
 print("\nThe stationary point is a Local Minimum.")
elif D > 0 and fxx < 0:
  print("\nThe stationary point is a Local Maximum.")
elif D < 0:
  print("\nThe stationary point is a Saddle Point.")
else:
  print("\nTest is Inconclusive.")
```
